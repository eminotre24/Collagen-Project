```python
import numpy as np
import matplotlib.pyplot as plt

# Line to unpack .xvg generated by GROMACS)
t_p, pres = np.loadtxt("./pres.xvg",comments=["@","#"],unpack=True)
t_e, ene = np.loadtxt("./energy.xvg",comments=["@","#"],unpack=True)

first_step = 10 # Given enourmous size of system
```

```python
plt.figure(dpi=200)
plt.plot(t_p[first_step:-1] ,pres[first_step:-1], label="pressure")
plt.xlabel("step")
plt.ylabel("bar")
plt.title("Evolution of Pressure - MD Run")
plt.legend()
plt.show()
```

```python
plt.figure(dpi=200)
plt.plot(t_e[first_step:-1], ene[first_step:-1], label="energy")
plt.xlabel("step")
plt.ylabel("kJ")
plt.title("Evolution of Energy - MD Run - U + K")
plt.legend()
plt.show()
```
