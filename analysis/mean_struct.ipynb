# Code for routine of generating mean fibril data

```python
import MDAnalysis as md
import numpy as np
import twistaly as tw

parent_dir = "data/"
file_dir = "data100ns-2008"
output_dir = "output-100ns2008"

u = md.Universe(parent_dir + file_dir + "/md_1.tpr", parent_dir + file_dir + "/md_1_noPBC.xtc")
prot = u.select_atoms("protein")
u.trajectory[-1]

t_eq = 20000 # ps
# Time Data
ti = u.trajectory[0].time
tf = u.trajectory[-1].time
nt = len(u.trajectory)-1
deltat = (tf-ti)/nt
t = np.linspace(ti,tf,nt+1)
maskt = (t >= t_eq)
```

## Generate Mean Fibril

```python
u_write = tw.mean_coords_generation(u, prot, maskt)
```

## Save the Data

```python
# PDB output
pdb = False

if pdb:
    with md.Writer(parent_dir + output_dir + "/meansys.pdb") as pdb:
        pdb.write(u_write)

# XTC for further Analysis
with md.Writer(parent_dir + output_dir + '/md_1_mean.xtc', n_atoms=u_write.atoms.n_atoms) as W:
    W.write(u_write.atoms) 
```
