```python
# We go through each chain of the molecule, check if its in the range, and if it is we save it in our new structure

import MDAnalysis as mda

# Load the full collagen structure
u = mda.Universe("colmol.pdb")

# Define Z limits (in Ångstroms)
z_min = 0
z_max = 720

# Find chains fully contained in the Z range
chains_to_keep = []
for seg in u.segments:
    z_coords = seg.atoms.positions[:, 2]  # z is the third column
    if z_coords.min() >= z_min and z_coords.max() <= z_max:
        chains_to_keep.append(seg)

# Select atoms from these chains
selected_atoms = u.select_atoms("segid " + " ".join(seg.segid for seg in chains_to_keep))

# Write to new file
with mda.Writer("colmol_trimmed.pdb", selected_atoms.n_atoms) as W:
    W.write(selected_atoms)
```
