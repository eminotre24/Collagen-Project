```python
import MDAnalysis as md

name = "colmol"

# Load structure
u = md.Universe(name + ".pdb")

zmin = 100 # 10 nm
zmax = 400

# Account for not centered object
basez = min(u.atoms.positions[:, 2])
zmin = basez + zmin
zmax = basez + zmax

# Get atoms from residues fully inside the Z-range
atoms_to_keep = []
for res in u.residues:
    z = res.atoms.positions[:, 2]
    if z.min() >= zmin and z.max() <= zmax:
        atoms_to_keep.extend(res.atoms.indices)

# Convert to AtomGroup
sel_atoms = u.atoms[atoms_to_keep]

output_name = name + "_cutd" + ".pdb"

# Write to PDB
with md.Writer(output_name , sel_atoms.n_atoms) as w:
    w.write(sel_atoms)

length_nm = (max(u.atoms.positions[:, 2]) - min(u.atoms.positions[:, 2]))/10
print("Original Length (nm): ", length_nm)

lengthf_nm = (max(sel_atoms.positions[:, 2]) - min(sel_atoms.positions[:, 2]))/10
print("Final Length (nm): ", lengthf_nm)
```

Seems to have worked for generating the `.gro` file through the command `pdb2gmx`. Its based of not breaking the residues, as we select all the atoms which consititute each residue which is in the range which we indicated.
