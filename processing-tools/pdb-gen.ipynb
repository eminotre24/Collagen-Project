```python
import MDAnalysis as md
import numpy as np

univ = md.Universe("md_1.tpr", "md_1_noPBC.xtc")
time = 20000

univ.trajectory[-1]
ti = univ.trajectory[0].time
tf = univ.trajectory[-1].time
nt = len(univ.trajectory)-1
deltat = (tf-ti)/nt
t = np.linspace(ti,tf,nt+1)
frame = int(time/deltat)

# Go to frame
univ.trajectory[frame]
prot = univ.select_atoms("protein")

with md.Writer(f"protframe{time/1e3:.0f}ns.pdb") as pdb:
    pdb.write(prot)
```
