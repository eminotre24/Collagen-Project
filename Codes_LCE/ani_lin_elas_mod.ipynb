### Anisotropic Linear Elasticity
Now I follow some of the calculations that Gouws did, in the part of the relations of the Young Moduli, and the Poisson Ratio. He made the following assumptions:
 - Under small strains: A linear elasticity is considered.
 - Cylindrical Symmetries: As the model is a cylinder, theres an important symmetry in $r:x,y$.
 - No shear: Since there is no shear under hydration, we can discard terms involving this shear deformations, i.e., no $\sigma_{ij}, i\neq j$.

The compliance matrix ultimately gives:

$$
\left[\begin{matrix} \epsilon_r \\ \epsilon_z \end{matrix}\right] = \left[\begin{matrix} (1-\nu_{rr'})/E_r && -\nu_{zr}/E_z \\ -2\nu_{rz}/E_r && 1/E_z \end{matrix}\right] \left[\begin{matrix} \sigma_r \\ \sigma_z \end{matrix}\right]
$$

So the stress matrix, or the stress equations can be given by the inverse of this matrix, we can try and compute the inverse using *SymPy*.

```python
import sympy as sp

nu_rr, nu_rz, nu_zr, E_r, E_z = sp.symbols('nu_rr nu_rz nu_zr E_r E_z')

compmat = sp.Matrix([[(1-nu_rr)/E_r, -nu_zr/E_z],[-2*nu_rz/E_r, 1/E_z]])

inv_comp = compmat.inv()

inv_comp
```

For which in fact we get the expression for the stresses. The expression of the stresses, as a function of constants: $\{E\}, \{\nu\}$, and the vector $\vec{\epsilon}, \vec{\sigma}(\vec{\epsilon})$, is useful as the strains depend upon the parameters of elongation, $\lambda_r, \lambda_z$.

Also and most importantly, as the stresses can relate to the potential energy of the system, which then we know minimizes under equilibrium. The general expression of this potential energy, for **small strains**, per unit volume is:

$$
u = \frac{1}{2}\Sigma_{i,j}\epsilon_{ij}\sigma_{ij}
$$

So we have $\sigma(\epsilon), \epsilon_i = \lambda_i - 1$, and then with the $\lambda$ s we can pass to the swelling ratio $\beta$ and deformation anisotropy $\alpha$. And minimize with respect of the deformation anisotropy, $\alpha$.
