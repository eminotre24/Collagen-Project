```python
import numpy as np
import matplotlib.pyplot as plt

# Trying to plot the direction of the azimuthal unit vector
n = 15
x = np.linspace(-20, 20, n)
y = np.linspace(-20, 20, n)

# Make a meshgrid so that then we can get the field
xv, yv = np.meshgrid(x, y)

theta = np.arctan2(yv, xv)

theta_x = -np.sin(theta)
theta_y = np.cos(theta)
```

```python
# Plot the vector field - quiver
plt.figure(dpi=150)
plt.quiver(xv, yv, theta_x, theta_y)
plt.title(r"$\hat{\phi}$" + " direction throught the plane" + r" $(x,y)$")
plt.xlabel(r"$x$")
plt.ylabel(r"$y$")
plt.show()
```

```python
# Now try the radial vector

rho_x = np.cos(theta)
rho_y = np.sin(theta)

plt.figure(dpi=150)
plt.quiver(xv, yv, rho_x, rho_y)
plt.title(r"$\hat{\rho}$" + " direction throught the plane" + r" $(x,y)$")
plt.xlabel(r"$x$")
plt.ylabel(r"$y$")
plt.show()
```

```python
# Superpose the 2 to make see the orthogonality

plt.figure(dpi=300)
plt.quiver(xv, yv, rho_x, rho_y, color="blue", label = "radius")
plt.quiver(xv, yv, theta_x, theta_y, color="red", label = "angle")
plt.title(r"$\hat{\rho}$" + " and " + r"$\hat{\phi}$" + " direction throught the plane" + r" $(x,y,0)$")
plt.xlabel(r"$x$")
plt.ylabel(r"$y$")
plt.legend()
plt.show()
```

Twisted Field Direction Vector:

$$
\hat{\textbf{n}} = -\sin \psi (r) \hat{\textbf{e}}_\phi + \cos \psi (r) \hat{\textbf{e}}_z
$$

```python
# Now graph the n vector of the twisted field for given the function

# Set a twist angle as constant: Psi_0, and we fix a cartesian coord so to have just 2 dimensions (atm)

psi_0 = np.deg2rad(40)
n = 10

x = np.linspace(-40, 40, n)
y = 5
z = np.linspace(0.1, 40, n)

xv, zv = np.meshgrid(x, z)

theta = np.arctan2(y, xv)
theta_x = -np.sin(theta)
theta_y = np.cos(theta)

norm_t = np.sqrt(theta_x**2 + theta_y**2)
theta_x = theta_x/norm_t
theta_y = theta_y/norm_t

z_tilt = np.cos(psi_0)*zv/abs(zv)

theta_x_tilt = -np.sin(psi_0)*theta_x
```

```python
# Check this calculation of the magnitude as it may not be correct.

plt.figure(dpi=300)
plt.quiver(xv, zv, theta_x_tilt, z_tilt)
plt.title(r"$\hat{n}$" + " direction throught the plane" + r" $(x,z)$")
plt.xlabel(r"$x$")
plt.ylabel(r"$z$")
plt.show()
```

```python
# 3D Visualization ? 
psi_0 = np.deg2rad(40)

first = -20
last = 20
n = 4

x, y, z = np.meshgrid(np.linspace(first, last, n),
                      np.linspace(first, last, n),
                      np.linspace(first, last, n))

theta =  np.arctan2(y, x)

theta_x = -np.sin(theta)
theta_y = np.cos(theta)

theta_x_tilt = -np.sin(psi_0)*theta_x
theta_y_tilt = np.cos(psi_0)*theta_y

z_tilt = np.cos(psi_0)*z/z
# No contribution in the radial part
```

```python
ax = plt.figure(dpi=300).add_subplot(projection='3d')
ax.quiver(x, y, z, theta_x_tilt, theta_y_tilt, z_tilt, length=6, normalize=1)
ax.axes.set_xlim3d(left=first*2, right=last*2) 
ax.axes.set_ylim3d(bottom=first*2, top=last*2) 
ax.axes.set_zlim3d(bottom=first*2, top=last*2) 
plt.show()
```
