### DATA READING

Having the Gouws data and some of the theoretical/equations background of the Hydration Project, then we can figure out the relationships highlighted or looked at.

Notes: RH stands for relative humidity, Total per dry (change in mass?) - saturation is $\text{T}_{pd}/67$ (Qualitative?)


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Read Haverkamp data - from final data one
coldat = pd.read_csv('./exp_data_gouws_fin/haverkamp_data.csv')
```

Now with this data lets try to get the same graphs as the paper has.

```python
raddef = np.array(coldat["radial deformation"].tolist())
axdef = np.array(coldat["axial deformation"].tolist())

# Plot the figure 4.1 winter thesis - (beta, alpha)

alpha = raddef/axdef
beta = raddef**2 * axdef

plt.figure(dpi=300)
plt.scatter(beta, alpha, color='g', marker='^')
plt.xlabel(r'$\beta$')
plt.ylabel(r'$\alpha$')
plt.xlim(0.38,1.35) # approx visually
plt.ylim(0.6,1.2) # approx visually
plt.title("Figure from data of Gouws - " + r"$(\beta,\alpha)$")
plt.show()
```
