### Free Energy modeling as a function of different parameters

As of the second day of work, we have the following structure/approach to the problem proposed here. We have a **double twist elastomer**, which means that we have the following considerations for the model of our collagen. We have what is called a **liquid crystal elastomer** (LCE), which consist of a **polymer** (an elastomer one, which has crosslinks), which has attached **mesogens** (molecules with anisotropy) which give the material certain crystal properties.

So first it is defined a directional field, which is the arrangement that the molecules take in the liquid crystal, in a directional way (orientation). So, in our case we have a double twist elastomer, which:

$$
\hat{\textbf{n}} = -\sin(\psi(r))\hat{\textbf{e}}_\phi + \cos(\psi(r))\hat{\textbf{e}}_z
$$

In which the parameter $\psi(r)$ represents the twist angle, sometimes dependant of the angle, sometimes constant, depends on the material. So, with our directional field for this model, we can define the **step-length tensor**, which is the arrangement that the mesogens, *in average*, take throughout the polymer/elastomer. This comes from the statistical theory of poluymer chains and networks, in which the polymer is considered as a random walk, and so in average it would behave a certain way. With the directional field vector, this tensor is then defined as:

$$
\textbf{L} = \ell^\perp\textbf{I}+(\ell^{||}-\ell^\perp)\hat{\textbf{n}}\otimes\hat{\textbf{n}}
$$

Where $\ell^\perp$ is the step-length scale in the perpendicular axis to the directional field, and $\ell^{||}$ the parallel one. This then yields in an step-length tensor that tells us the average anisotropic shape of polymer coils in space, that is how our LCE is in general arranged, consistent somewhat with what the directional field describes. An important parameter that is used later in the works of the research group is the anisotropy parameter, $\zeta$, as;

$$
\zeta = \frac{\ell^{||}}{\ell^\perp}
$$

With this we have a general mechanical description of our LCE model, given its characteristics. Now, as we study the deformation of it, we need the following physical tools. We need the deformation tensor, which is denotated by $\bold{\lambda}$ or $\bold{\Lambda}$, and its defined as the type of deformation we are looking at. So, for example, for hydratation and dehydratation, we have a radial and $z$-axis deformation due to the sweeling, in which cases they expand (not fixed volume). There are other cases in which the volume is constant, and the deformation is of enlongation, in which we have $\bold{\Lambda} = \text{diag}(1/\sqrt{\lambda},1/\sqrt{\lambda},\lambda)$. Here arise other parameters that are used through the investigation, which are the following:

- Swelling Ratio: $\beta = \lambda_r^2\lambda_z = V/V_0$
- Deformation Anisotropy: $\alpha = \lambda_r/\lambda_z$

With this tensor, then we can define that the **free energy density** under this circumstances is then:

$$
f = \frac{\mu}{2}\left\{\text{Tr}(\textbf{L}_0\bold{\Lambda}^{\text{T}}\textbf{L}^{-1}\bold{\Lambda}) + \ln{\left[\frac{\det{(\bold{L})}}{\det{(\bold{L_0})}}\right]}\right\}
$$

This is the general form given by the LCE theory, which is used throughout the different texts (Molecular Theory of LCE, Nematic Elastomers). So, with this, theres really a broad avenue of exploration that is made, and that also can be made, with different considerations made, the different type of behavior/effects that want to be modeled, and the parameter that wants to be observerd.

So, I think we can work a simple case to start and make some graphs, compare the results with the ones of the papers, and reinforce the understanding.

```python
# Get the term Tr() - Nematic Configuration psi = 0
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt

lambdar, lambdaz, ell0_perp, ell0_par, ell_perp, ell_par = sp.symbols('lambda_r lambda_z ell_0^perp ell_0^par ell^perp ell^par')

L_0 = sp.Matrix([[ell0_perp, 0, 0], [0, ell0_perp, 0], [0, 0, ell0_par]])

L = sp.Matrix([[ell_perp, 0, 0], [0, ell_perp, 0], [0, 0, ell_par]])

Lambda = sp.Matrix([[lambdar, 0, 0], [0, lambdar, 0], [0, 0, lambdaz]])

trace_arg = L_0 * sp.Transpose(Lambda) * sp.Inverse(L) * Lambda

trace = sp.Trace(trace_arg).rewrite(sp.Sum)

trace
```

Derive the free energy minimized expression simplified (due to $\beta, \mu \neq 0$):

$$
0=\frac{\partial}{\partial\alpha}\left[ \frac{2\ell_0^\perp}{\ell^\perp}\alpha^{2/3} + \frac{\ell^{||}_0}{\ell^{||}}\alpha^{-4/3} \right]
$$

```python
alpha = sp.symbols('alpha')

expr = alpha**sp.Rational('2/3')*(2*ell0_perp/ell_perp) + alpha**sp.Rational('-4/3')*(ell0_par/ell_par)

zeroexpr = sp.diff(expr, alpha)

solu = sp.solve(zeroexpr, alpha)

solu[1]
# zeroexpr
```

Based on whats was found to be the equilibrium anisotropic deformation, and $\zeta$, the anisotropic parameter defintion, we have:

$$
\alpha_{eq} = \sqrt{\frac{\ell^\perp\ell_0^{||}}{\ell^{||}\ell_0^\perp}}, \quad \zeta = \frac{\ell^{||}}{\ell^\perp}
$$

We would have that in this nematic limit, the equilibrium anisotropic deformation is then given as:

$$
\alpha_{eq} = \sqrt{\frac{\zeta_0}{\zeta}}
$$
