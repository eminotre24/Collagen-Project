```python
# Let us consider the professor's approach, a quite interesting deformation

import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# Def Tensor
lambda_r, lambda_z = sp.symbols('lambda_r lambda_z')
Lambda = sp.diag(lambda_r, lambda_r, lambda_z)

Id = sp.eye(3)

# Director field
xhat = sp.Matrix([1,0,0])
yhat = sp.Matrix([0,1,0])
zhat = sp.Matrix([0,0,1])

phi = sp.symbols('phi')

phihat = -sp.sin(phi)*xhat + sp.cos(phi)*yhat

psi_0, psi = sp.symbols('psi_0 psi')

nhat_0 = -sp.sin(psi_0)*phihat + sp.cos(psi_0)*zhat
nhat = -sp.sin(psi)*phihat + sp.cos(psi)*zhat

# Steplengths
zeta, zeta_0 = sp.symbols('zeta zeta_0')

Lt0 = Id + (zeta_0 - 1)*nhat_0*sp.transpose(nhat_0)

Lt = Id + (zeta - 1)*nhat*sp.transpose(nhat)
```

```python
# Free Energy
mu = sp.symbols('mu')

free_energy = mu/2 * (sp.trace(Lt0 * sp.transpose(Lambda) * Lt.inv() * Lambda) + sp.log(sp.det(Lt)/sp.det(Lt0)))

free_energy = free_energy.simplify()

free_energy = free_energy.trigsimp()

free_energy
```

```python
free_energy = free_energy.simplify()

free_energy = sp.collect(free_energy, [sp.sin(psi_0)**2 * sp.sin(psi)**2])

free_energy = free_energy.simplify()

free_energy = sp.collect(free_energy, [4*lambda_z**2, (sp.cos(2*psi - 2*psi_0) - sp.cos(2*psi + 2*psi_0))*lambda_r*lambda_z, 4*lambda_r**2, lambda_r*lambda_z*zeta*zeta_0])

free_energy
```

```python
free_energy = free_energy.simplify()
free_energy = free_energy.trigsimp()

free_energy
```

As far as we can get with the simplification. We have this free energy as a function of:

$$
f = f(\lambda_r, \lambda_z, \psi_0, \psi, \zeta_0, \zeta)
$$

In principle, we have the suposition that there could be the following relationship: $\ell(\lambda) \rightarrow \zeta(\lambda_r, \lambda_z)$, which lambda with which deformation is yet to be considered.

Since the approach we can see that the free energy we just stated will not behave as normally due, as the deformation is also in the step-length tensor, meaning that the deformation tensor doesnt encompass the whole deformation.

So we wanna found the **anisotropic deformation** which holds in equilibrium given our system, that is to minimize our free energy with respect to this parameter, which would give us the one system with enholds the equilibrium alpha.

We take the following relations by defintion:

$$
\lambda_r = \alpha^{1/3}\beta^{1/3}, \quad \lambda_z = \alpha^{-2/3}\beta^{1/3}
$$

And then take the derivative of the function.


```python
alpha, beta = sp.symbols('alpha beta')

free_energy_ab = free_energy.subs(lambda_r, alpha**sp.Rational('1/3')*beta**sp.Rational('1/3'))
free_energy_ab = free_energy_ab.subs(lambda_z, alpha**sp.Rational('-2/3')*beta**sp.Rational('1/3'))

free_energy_ab = free_energy_ab.simplify()

free_energy_ab
```

```python
# Take the derivative with respect of \alpha

df_da = sp.diff(free_energy_ab, alpha)

df_da = df_da.simplify()

df_da
```

```python
# Imagine we would want to take the nematic limit right now
df_da_nem = df_da.evalf(subs={psi_0:0, psi:0})

solu_nem = sp.solve(df_da_nem, alpha)

solu_nem[1]
```

This is concise with what we previously had obtained.

```python
# Brute-Force solve to find the equilibrium alpha

solu = sp.solve(df_da, alpha)

solu[1].simplify()
```

Small Angle Approximation

```python
# Maybe if we take an small angle approximation we will be able to get something more interesting -sa:small angle

df_da_sa = df_da.subs(sp.sin(psi), psi)
df_da_sa = df_da_sa.subs(sp.sin(psi_0), psi_0)
df_da_sa = df_da_sa.subs(sp.sin(psi)**2, psi**2)
df_da_sa = df_da_sa.subs(sp.sin(psi_0)**2, psi_0**2)

'''
df_da_sa = df_da_sa.subs(sp.cos(psi), 1 - psi**2/2)
df_da_sa = df_da_sa.subs(sp.cos(psi_0), 1 - psi_0**2/2)
df_da_sa = df_da_sa.subs(sp.cos(psi)**2, (1 - psi_0**2/2)**2)
df_da_sa = df_da_sa.subs(sp.cos(psi_0)**2, (1 - psi_0**2/2)**2)
'''
df_da_sa = df_da_sa.subs(sp.cos(2*psi - 2*psi_0), 1)
df_da_sa = df_da_sa.subs(sp.cos(2*psi + 2*psi_0), 1)


df_da_sa = df_da_sa.simplify()

df_da_sa
```

```python
solu_sa = sp.solve(df_da_sa, alpha)
```
